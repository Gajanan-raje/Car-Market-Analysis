```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")


df = pd.read_csv("1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv")

# Basic check
print("Shape:", df.shape)
df.head()
```

```python
print(df.info())
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Remove duplicates
df = df.drop_duplicates()
print("New shape after removing duplicates:", df.shape)

# Add Car Age column (dataset 2020 cha ahe assume karून)
current_year = 2020
df['Car_Age'] = current_year - df['Year']

df.head()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Selling_Price'], bins=25, kde=True, color="#2E86AB")
plt.title("Distribution of Selling Price (in Lakhs)")
plt.xlabel("Selling Price (Lakhs)")
plt.ylabel("Count of Cars")
plt.tight_layout()
plt.savefig("01_selling_price_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
order = df['Fuel_Type'].value_counts().index
sns.countplot(data=df, x='Fuel_Type', order=order, hue='Fuel_Type', palette="viridis", legend=False)
plt.title("Number of Cars by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig("02_fuel_type_count.png")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x='Transmission', hue='Transmission', palette="magma", legend=False)
plt.title("Number of Cars by Transmission Type")
plt.tight_layout()
plt.savefig("03_transmission_count.png")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='Fuel_Type', y='Selling_Price', hue='Fuel_Type', palette="Set2", legend=False)
plt.title("Selling Price by Fuel Type")
plt.ylabel("Selling Price (Lakhs)")
plt.tight_layout()
plt.savefig("04_price_by_fuel_type.png")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.boxplot(data=df, x='Transmission', y='Selling_Price', hue='Transmission', palette="Set3", legend=False)
plt.title("Selling Price by Transmission Type")
plt.ylabel("Selling Price (Lakhs)")
plt.tight_layout()
plt.savefig("05_price_by_transmission.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Car_Age', y='Selling_Price', hue='Fuel_Type', s=60, alpha=0.8)
plt.title("Car Age vs Selling Price")
plt.xlabel("Car Age (Years)")
plt.ylabel("Selling Price (Lakhs)")
plt.tight_layout()
plt.savefig("06_age_vs_price.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Kms_Driven', y='Selling_Price', hue='Fuel_Type', s=60, alpha=0.8)
plt.title("Kms Driven vs Selling Price")
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price (Lakhs)")
plt.tight_layout()
plt.savefig("07_kms_vs_price.png")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

df = pd.read_csv("1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv")
df = df.drop_duplicates()

current_year = 2020
df['Car_Age'] = current_year - df['Year']

print("Shape:", df.shape)
df.head()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Kms_Driven', y='Selling_Price', hue='Fuel_Type', s=60, alpha=0.8)
plt.title("Kms Driven vs Selling Price")
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price (Lakhs)")
plt.tight_layout()
plt.savefig("07_kms_vs_price.png")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.boxplot(data=df, x='Seller_Type', y='Selling_Price', hue='Seller_Type', palette="coolwarm", legend=False)
plt.title("Selling Price by Seller Type")
plt.ylabel("Selling Price (Lakhs)")
plt.tight_layout()
plt.savefig("08_price_by_seller_type.png")
plt.show()
```

```python

plt.figure(figsize=(7, 6))
num_cols = ['Selling_Price', 'Present_Price', 'Kms_Driven', 'Car_Age', 'Owner']
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.savefig("09_correlation_heatmap.png")
plt.show()
```

```python
print("=== KEY INSIGHTS ===")
print(f"Total cars analyzed: {len(df)}")
print(f"Average Selling Price: {df['Selling_Price'].mean():.2f} Lakhs")
print(f"Average Car Age: {df['Car_Age'].mean():.1f} years")
print(f"Most common Fuel Type: {df['Fuel_Type'].mode()[0]}")
print(f"Most common Transmission: {df['Transmission'].mode()[0]}")

diesel_avg = df[df.Fuel_Type=='Diesel']['Selling_Price'].mean()
petrol_avg = df[df.Fuel_Type=='Petrol']['Selling_Price'].mean()
print(f"\nDiesel avg price: {diesel_avg:.2f} Lakhs")
print(f"Petrol avg price: {petrol_avg:.2f} Lakhs")

auto_avg = df[df.Transmission=='Automatic']['Selling_Price'].mean()
manual_avg = df[df.Transmission=='Manual']['Selling_Price'].mean()
print(f"\nAutomatic avg price: {auto_avg:.2f} Lakhs")
print(f"Manual avg price: {manual_avg:.2f} Lakhs")

print(f"\nCorrelation (Present Price vs Selling Price): {corr.loc['Present_Price','Selling_Price']:.2f}")
print(f"Correlation (Car Age vs Selling Price): {corr.loc['Car_Age','Selling_Price']:.2f}")
```
